## 📋 The Problem

> A cookie recipe calls for the following ingredients:
> - 1.5 cups of sugar
> - 1 cup of butter
> - 2.75 cups of flour
>
> The recipe produces **48 cookies** with this amount of the ingredients.
>
> Write a program that **asks the user how many cookies** he or she wants to make. Then **display the number of cups of each ingredient** needed for the specified number of cookies.

## 🧠 Step 0: Think Before You Code (IPO Chart)

Every program does three things. Fill in this chart in your head (or on paper) before typing any code.

| 📥 **Input** | ⚙️ **Processing** | 📤 **Output** |
|---|---|---|
| What does the user give us? | What math do we do? | What do we show? |
| The number of cookies they want | Figure out how much bigger or smaller the batch is, then multiply each ingredient | Cups of sugar, butter, and flour |

**Big idea:** The original recipe makes 48 cookies. If someone wants **24** cookies, that is **half** a recipe, so every ingredient is cut in half. If someone wants **96** cookies, that is **double** a recipe, so every ingredient is doubled.

## Part 1: Store the Recipe in Variables

First, we save the facts from the recipe. These numbers will not change while the program runs.

💡 *Tip: ALL_CAPS names are a Python habit for values that stay the same.*

```python
# The ORIGINAL recipe (makes 48 cookies)
SUGAR_CUPS = 1.5
BUTTER_CUPS = 1
FLOUR_CUPS = 2.75
COOKIES_IN_RECIPE = 48

print("Sugar:", SUGAR_CUPS, "cups")
print("Butter:", BUTTER_CUPS, "cup")
print("Flour:", FLOUR_CUPS, "cups")
print("Makes:", COOKIES_IN_RECIPE, "cookies")
```

## Part 2: Processing — The Scale Factor

To change the recipe size, we need a **scale factor**:

`scale_factor = cookies_wanted / COOKIES_IN_RECIPE`

| Cookies wanted | Calculation | Scale factor |
|---|---|---|
| 48 | 48 / 48 | 1.0 (same recipe) |
| 24 | 24 / 48 | 0.5 (half) |
| 96 | 96 / 48 | 2.0 (double) |

Then each ingredient is: `new amount = original amount * scale_factor`

Let's try it with a **pretend** answer of 24 cookies (no typing needed yet):

```python
# Run the SAME recipe as in Part 1 first, then this cell.
cookies_wanted = 24          # pretend the user typed 24

scale_factor = cookies_wanted / COOKIES_IN_RECIPE
print("Scale factor:", scale_factor)

sugar_needed = SUGAR_CUPS * scale_factor
print("Sugar needed:", sugar_needed, "cups")
```

## Part 3: Input — Ask the User

The `input()` function pauses the program and waits for the user to type something.

⚠️ **Important:** `input()` always gives you **text** (a string), even if the user types a number. Math does not work on text, so we convert it with `int()`.

Run this cell to see the difference:

```python
answer = "24"                 # pretend this is what input() gave us
print(type(answer))           # <class 'str'>  (text)

cookies = int(answer)         # convert the text to a whole number
print(type(cookies))          # <class 'int'>  (number)

print(cookies * 2)            # math works on numbers
```

Now try the real thing. Run this cell, type a number into the box, and press **Enter**:

```python
cookies_wanted = int(input("How many cookies do you want to make? "))
print("You want to make", cookies_wanted, "cookies!")
```

## Part 4: Output — Show the Results Nicely

We can print a message with the values inside it using an **f-string** (put an `f` before the quotes and place variables inside `{ }`).

We can also **round** to 2 decimal places with `:.2f`. Look at the difference:

```python
amount = 1.375

print("Without formatting:", amount)
print(f"With formatting: {amount:.2f} cups")
```

## 🔑 Part 7: Sample Solution

*Try it on your own first! Then compare your work with this solution. Your code can look different and still be correct.*

```python
# ---------- Cookie Recipe Calculator ----------

# Input data: the original recipe (makes 48 cookies)
SUGAR_CUPS = 1.5
BUTTER_CUPS = 1
FLOUR_CUPS = 2.75
COOKIES_IN_RECIPE = 48

# Input: ask the user
cookies_wanted = int(input("How many cookies do you want to make? "))

# Processing: scale the recipe up or down
scale_factor = cookies_wanted / COOKIES_IN_RECIPE
sugar_needed = SUGAR_CUPS * scale_factor
butter_needed = BUTTER_CUPS * scale_factor
flour_needed = FLOUR_CUPS * scale_factor

# Output: show the results
print()
print(f"To make {cookies_wanted} cookies you will need:")
print(f"  Sugar:  {sugar_needed:.2f} cups")
print(f"  Butter: {butter_needed:.2f} cups")
print(f"  Flour:  {flour_needed:.2f} cups")
```
