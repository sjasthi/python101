#L. Bitwise Operators

```python
#WWPP L.1
#hex #oct #scientific notation
a, b, c, d = 0, 1, 0, 1
p =  (a & b)
q =  (b | c)
r =  (b ^ d)

print(p, q, r,  sep = '')
```

```python
#WWPP L.2
a = 2
x = a << 1
y = a << 2
print(x+y)
```

```python
#WWPP L.3
a = 32
x = a >> 1
y = x >> 1
print(x+y)
```

```python
#WWPP L.4
a = '0b1111'
x = int(a,2)
print(x)
```

```python
#WWPP L.5
a = 0b1011
b = 0b1101
x = a | b
print(x)
```

```python
#WWPP L.6
a = 0b1011
b = 0b1101
c = a | b
x = bin(c)
x = x[2:] # Remove the 'ob' prefix
print(x)
```
